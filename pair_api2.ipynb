```python
headers = {
    "Authorization": "54a45d80-2222-47f0-b15d-bb8c7fcd9d41"
}

url = "https://api.balldontlie.io/v1/players"

parametros = {
    "per_page": 100
}

jugadores = []
total = 0
cursor = None

```

```python
while total < 200:

    if cursor:
        parametros["cursor"] = cursor

    response = requests.get(url, headers=headers, params=parametros)
    diccplayers = response.json()

    for player in diccplayers["data"]:

        jugadores.append({
            "nombre": player["first_name"],
            "apellido": player["last_name"],
            "equipo": player["team"]["full_name"]
        })

        total += 1

        if total >= 200:
            break

    cursor = diccplayers["meta"]["next_cursor"]

df = pd.DataFrame(jugadores)

print(df.head())
print("Total jugadores:", len(df))
```
