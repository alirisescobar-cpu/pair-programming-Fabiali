```python
import requests

def buscar_personaje():
    print("--- BUSCADOR DE STAR WARS ---")
    busqueda = input("Escribe el nombre del personaje: ")
    
    url = "https://swapi.py4e.com/api/people/"
    params = {'search': busqueda}
    
    try:
        res = requests.get(url, params=params)
        data = res.json()
        results = data.get('results', [])
        
        if not results:
            print("No hubo suerte. ¡Prueba con otro nombre!")
            return

        for p in results:
            print(f"-> {p['name']} ({p['gender']})")
            
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

# Ejecutamos la función
buscar_personaje()
```
